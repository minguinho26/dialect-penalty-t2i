```python
import json, random
random.seed(42)
```

```python
with open('./T2I-RiskyPrompt.json', 'r') as f:
    prompts = json.load(f)

prompts_dict = {}

for sample_idx in prompts:

    prompt, label = prompts[sample_idx]['prompt'], prompts[sample_idx]['label']

    primary_label = list(prompts[sample_idx]['label'].keys())[0]

    sub_category = primary_label.split('--')[-1]

    if sub_category not in prompts_dict:
        prompts_dict[sub_category] = []

    prompts_dict[sub_category].append(prompts[sample_idx])

with open('./T2I-RiskyPrompt_per_category.json', 'w') as f:
    json.dump(prompts_dict, f, indent=4, ensure_ascii=False)
```
