```python
import pandas as pd

def print_wide_latex_results(csv_path="sweep_table.csv"):
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"File '{csv_path}' not found. Check the path!")
        return

    df['mean_pct'] = df['mean'] * 100
    df['std_pct'] = df['std'] * 100

    df['latex_format'] = df.apply(
        lambda row: f"${row['mean_pct']:.3f} \\pm {row['std_pct']:.3f}$", 
        axis=1
    )

    df_pivot = df.pivot(
        index=['sae_frac', 'method'], 
        columns='metric', 
        values='latex_format'
    ).reset_index()

    df_pivot = df_pivot.sort_values(by=['sae_frac', 'method'], ascending=[True, False])

    print(f"{'SAE_FRAC':<10} | {'METHOD':<10} | {'TEST_ACCURACY':<22} | {'TEST_WORST_GROUP_ACC':<22}")
    print("-" * 75)
    
    for _, row in df_pivot.iterrows():
        sae = str(row['sae_frac'])
        method = str(row['method'])
        
        # Fallback to 'N/A' if metric is missing
        acc = str(row.get('test_accuracy', 'N/A'))
        worst = str(row.get('test_worst_group_acc', 'N/A'))
        
        print(f"{sae:<10} | {method:<10} | {acc:<22} | {worst:<22}")
```

```python
print_wide_latex_results('./results/sweep_ratio/sweep_table.csv')
```
