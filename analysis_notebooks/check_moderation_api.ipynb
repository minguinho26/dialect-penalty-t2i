```python
import pandas as pd
import os
```

```python
for dialect in ["AAVE", "ChcE", "CollSgE", "IndE", "JamE"]:

    csv_path = f'openai_moderation_results/{dialect}_toxic.csv'

    if os.path.exists(csv_path) == False:
        continue

    df = pd.read_csv(csv_path)

    result = df.groupby('category').apply(
        lambda x: pd.Series({
            'std=T & dial=T': ((x['standard_is_harmful'] == True) & (x['dialect_is_harmful'] == True)).sum(),
            'std=T & dial=F': ((x['standard_is_harmful'] == True) & (x['dialect_is_harmful'] == False)).sum(),
            'std=F & dial=T': ((x['standard_is_harmful'] == False) & (x['dialect_is_harmful'] == True)).sum(),
            'std=F & dial=F': ((x['standard_is_harmful'] == False) & (x['dialect_is_harmful'] == False)).sum()
        }),
        include_groups=False
    )

    result['total'] = result.sum(axis=1)

    print(f"results ({dialect}):")
    total_prompts = result['total'].sum()

    tpr_sae = (result['std=T & dial=T'].sum() + result['std=T & dial=F'].sum()) / total_prompts
    tpr_dial = (result['std=T & dial=T'].sum() + result['std=F & dial=T'].sum()) / total_prompts

    delta_tpr = tpr_dial - tpr_sae

    print(f"TPR_SAE:   {tpr_sae * 100:.3f}%")
    print(f"TPR_dial:  {tpr_dial * 100:.3f}%")
    print(f"ΔTPR:      {delta_tpr * 100:+.3f} pp")

    print("==================================")
```
