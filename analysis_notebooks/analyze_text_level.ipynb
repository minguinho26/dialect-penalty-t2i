```python
clip_sim = {}

for prompt_type in ['toxic', 'benign']: # 'toxic', 
    clip_sim[prompt_type] = {}
    for dialect in aggregate['Dialect'].to_list():
        df = pd.read_csv(f'./text_level_typo_results/{prompt_type}_results_{dialect}_seed_0.csv')
        clip_sim[prompt_type][dialect] = round(df['target_sim'].mean(), 3)
```
